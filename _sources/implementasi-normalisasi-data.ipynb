## 3. Implementasi Normalisasi Data (Tugas Code)

Bagian ini mendemonstrasikan hasil praktik kode (*code experimen*) untuk menormalisasi data menggunakan ketiga metode yang telah dijelaskan pada sub-bab teori: **Min-Max**, **Z-Score**, dan **Decimal Scaling**.

Kita akan meracik perbandingan hasil antara membuat fungsi kalkulasi formula murni dengan Python (*from scratch*) berhadapan dengan output hasil komputasi otomatis bawaan alat prapemrosesan terkemuka: `scikit-learn`.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 1. Membuat Dummy Dataset
data_karyawan = {
    'Umur': [25, 30, 48, 22, 53, 28],
    'Gaji_Pokok': [5000000, 8000000, 15000000, 4500000, 22000000, 6000000]
}
df = pd.DataFrame(data_karyawan)

print("--- Data Sebelum Normalisasi ---")
display(df)
```

### 3.1 Min-Max Normalization
Mentransformasikan data Gaji dan Umur agar memiliki bentangan rasio ketat di antara skala `0` hingga `1`.

```python
# Fungsi Manual (From Scratch) Min-Max
def manual_min_max_scaler(kolom):
    return (kolom - kolom.min()) / (kolom.max() - kolom.min())

# Eksekusi Manual
df_minmax_manual = df.copy()
df_minmax_manual['Umur'] = manual_min_max_scaler(df['Umur'])
df_minmax_manual['Gaji_Pokok'] = manual_min_max_scaler(df['Gaji_Pokok'])

# Eksekusi Menggunakan Library [Sklearn]
scaler_mm = MinMaxScaler()
df_minmax_sklearn = pd.DataFrame(scaler_mm.fit_transform(df), columns=df.columns)

print("--- Hasil Komputasi Manual Min-Max ---")
display(df_minmax_manual)

print("\n--- Hasil Sklearn MinMaxScaler ---")
display(df_minmax_sklearn)
```

### 3.2 Z-Score Standardization
Pemusatan distribusi Gaji dan Umur menuju kurva nol (Mean=0) diiringi rentang standar deviasi satu.

```python
# Fungsi Manual (From Scratch) Z-Score
def manual_z_score_scaler(kolom):
    # Populasi standar deviasi menggunakan `ddof=0` sama seperti parameter bawaan Sklearn
    return (kolom - kolom.mean()) / kolom.std(ddof=0)

# Eksekusi Manual
df_zscore_manual = df.copy()
df_zscore_manual['Umur'] = manual_z_score_scaler(df['Umur'])
df_zscore_manual['Gaji_Pokok'] = manual_z_score_scaler(df['Gaji_Pokok'])

# Eksekusi Menggunakan Library [Sklearn]
scaler_z = StandardScaler()
df_zscore_sklearn = pd.DataFrame(scaler_z.fit_transform(df), columns=df.columns)

print("--- Hasil Komputasi Manual Z-Score ---")
display(df_zscore_manual)

print("\n--- Hasil Sklearn StandardScaler ---")
display(df_zscore_sklearn)
```

### 3.3 Decimal Scaling
Penyesuaian pergeseran rentang nominal titik desimal.

```python
# Fungsi Manual (From Scratch) Decimal Scaling
def manual_decimal_scaling(kolom):
    # Mengambil nilai absolut terbesar dan menghitung jumlah panjang digitnya
    max_absolut = kolom.abs().max()
    digit_j = len(str(int(max_absolut)))
    # Penggeseran titik dengan pembagi kelipatan 10 berdasar penentuan j (10^j)
    return kolom / (10 ** digit_j)

# Catatan: Scikit-learn tidak mendistribusikan library bawaan untuk pergerakan desimal murni,
# Sehingga performa perumusan custom akan dieksekusi independen.
df_decimal_manual = df.copy()
df_decimal_manual['Umur'] = manual_decimal_scaling(df['Umur'])
df_decimal_manual['Gaji_Pokok'] = manual_decimal_scaling(df['Gaji_Pokok'])

print("--- Hasil Komputasi Manual Decimal Scaling ---")
display(df_decimal_manual)
```
